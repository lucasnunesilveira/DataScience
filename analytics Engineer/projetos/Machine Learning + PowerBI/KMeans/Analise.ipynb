```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

```

```python
df_Clientes = pd.read_csv('./Dados/dados_clientes.csv', sep=',')
df_Clientes.head(10)
```

```python
df_Clientes.info()
```

```python
df_Clientes.describe().round(2)
```

```python
padronizador = StandardScaler() #padroniza deixa mesma escala
df_Clientes_padronizado = padronizador.fit_transform(df_Clientes[['idade', 'renda_anual', 'pontuacao_gastos']])
print(df_Clientes_padronizado[:10])
```

```python
k = 3 #número de clusters
kmeans = KMeans(n_clusters=k) #cria o modelo KMeans com k clusters
kmeans.fit(df_Clientes_padronizado) #ajusta o modelo aos dados padronizados
df_Clientes['cluster'] = kmeans.labels_ #adiciona os rótulos dos clusters ao DataFrame original
```

```python
df_Clientes.head(10)
```

```python
df_Clientes.to_csv('./Dados/dados_clientes_clusters.csv', sep=',', index=False)
```
