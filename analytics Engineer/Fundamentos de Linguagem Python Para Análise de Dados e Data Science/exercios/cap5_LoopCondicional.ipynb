## Exercícios - Loops e Condiconais

```python
# Exercício 1 - Crie uma estrutura que pergunte ao usuário qual o dia da semana. Se o dia for igual a Domingo ou 
# igual a sábado, imprima na tela "Hoje é dia de descanso", caso contrário imprima na tela "Você precisa trabalhar!"
dia = input('Qual o dia da semana? ').strip().lower()
if dia == 'domingo' or dia == 'sabado':
    print('Hoje é dia de descanso')
```

```python
# Exercício 2 - Crie uma lista de 5 frutas e verifique se a fruta 'Morango' faz parte da lista
frutas = ['maçã', 'banana', 'laranja', 'uva', 'morango']
if 'morango' in frutas:
    print('Morango está na lista de frutas')
```

```python
# Exercício 3 - Crie uma tupla de 4 elementos, multiplique cada elemento da tupla por 2 e guarde os resultados em uma 
# lista
tupla = (1, 2, 3, 4)
lista = [elemento * 2 for elemento in tupla]
print('Lista com elementos da tupla multiplicados por 2:', lista)
```

```python
# Exercício 4 - Crie uma sequência de números pares entre 100 e 150 e imprima na tela
pares = [num for num in range(100, 151) if num % 2 == 0]
print('Números pares entre 100 e 150:', pares)
```

```python
# Exercício 5 - Crie uma variável chamada temperatura e atribua o valor 40. Enquanto temperatura for maior que 35, 
# imprima as temperaturas na tela
temperatura = 40
while temperatura > 35:
    print('Temperatura atual:', temperatura)
    temperatura -= 1  # Decrementa a temperatura para evitar loop infinito
```

```python
# Exercício 6 - Crie uma variável chamada contador = 0. Enquanto counter for menor que 100, imprima os valores na tela,
# mas quando for encontrado o valor 23, interrompa a execução do programa
contador = 0
while contador < 100:
    if contador == 23:
        print('Contador atingiu o valor 23, interrompendo a execução.')
        break
    contador += 1  
print('Contador atual:', contador)   
```

```python
# Exercício 7 - Crie uma lista vazia e uma variável com valor 4. Enquanto o valor da variável for menor ou igual a 20, 
# adicione à lista, apenas os valores pares e imprima a list
lista_vazia = [] 
valor = 4
while valor <= 20:
    if valor % 2 == 0:
        lista_vazia.append(valor)
    valor += 1
print('Lista de valores pares entre 4 e 20:', lista_vazia)   
```

```python
# Exercício 8 - Transforme o resultado desta função range em uma lista: range(5, 45, 2)
nums = range(5, 45, 2)
lista_nums = list(nums)
print('Lista gerada a partir do range(5, 45, 2):', lista_nums)
```

```python
# Exercício 9 - Faça a correção dos erros no código abaixo e execute o programa. Dica: são 3 erros.
temperatura = input('Qual a temperatura? ')
temperatura = int(temperatura)

if temperatura > 30:
    print('Vista roupas leves.')
else:
    print('Busque seus casacos.')

```

```python
# Exercício 10 - Faça um programa que conte quantas vezes a letra "r" aparece na frase abaixo. Use um placeholder na 
# sua instrução de impressão

# “A gratidão é a virtude das almas nobres. O quão feliz é uma pessoa depende da profundidade de sua gratidão.” 

frase = "A gratidão é a virtude das almas nobres. O quão feliz é uma pessoa depende da profundidade de sua gratidão." 
contador_r = frase.lower().count('r')
print('A letra "r" aparece {} vezes na frase.'.format(contador_r))
```
